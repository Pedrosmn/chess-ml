```python
from pathlib import Path
import pandas as pd
import json

def load_chesscom_games(base_path="../data/raw"):
    all_games = []

    base_path = Path(base_path)

    # percorre todas as subpastas
    for game_type_path in base_path.iterdir():

        # ignora arquivos
        if not game_type_path.is_dir():
            continue

        game_type = game_type_path.name

        # percorre os jsons
        for json_file in game_type_path.glob("*.json"):

            with open(json_file, "r", encoding="utf-8") as f:
                data = json.load(f)

            # alguns endpoints retornam {"games": [...]}
            games = data.get("games", data)

            # transforma em df
            df = pd.DataFrame(games)

            # adiciona metadados
            df["game_type"] = game_type
            df["source_file"] = json_file.name

            all_games.append(df)

    # junta tudo
    final_df = pd.concat(all_games, ignore_index=True)

    return final_df
```
